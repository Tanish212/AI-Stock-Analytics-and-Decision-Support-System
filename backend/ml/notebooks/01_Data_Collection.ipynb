```python
import yfinance as yf
import pandas as pd
import os
```

```python
stocks = [
    "SBIN.NS",
    "RELIANCE.NS",
    "TCS.NS",
    "INFY.NS",
    "HDFCBANK.NS",
    "ICICIBANK.NS",
    "ITC.NS",
    "LT.NS",
    "AXISBANK.NS",
    "BHARTIARTL.NS"
]

START_DATE = "2015-01-01"
```

```python
RAW_DATA_PATH = "../data/raw"

for stock in stocks:

    print(f"Downloading {stock}...")

    df = yf.download(
        stock,
        start=START_DATE,
        auto_adjust=False,
        progress=False,
        multi_level_index=False
    )

    if not df.empty:

        file_name = stock.replace(".NS", "")

        df.reset_index(inplace=True)

        df.to_csv(
            f"{RAW_DATA_PATH}/{file_name}.csv",
            index=False
        )

        print(f"✓ Saved {file_name}.csv")

    else:
        print(f"✗ No data found for {stock}")
```

```python
import os
import pandas as pd

RAW_DATA_PATH = "../data/raw"

summary = []

for file in os.listdir(RAW_DATA_PATH):
    
    if file.endswith(".csv"):
        
        stock_name = file.replace(".csv", "")
        
        df = pd.read_csv(
            os.path.join(RAW_DATA_PATH, file)
        )
        
        summary.append({
            "Stock": stock_name,
            "Rows": len(df),
            "Start Date": df["Date"].min(),
            "End Date": df["Date"].max(),
            "Missing Values": df.isnull().sum().sum(),
            "Columns": len(df.columns)
        })

summary_df = pd.DataFrame(summary)

summary_df
```
