```python
import os

DATA_FOLDER = "../data/processed"

existing_files = [
    f for f in os.listdir(DATA_FOLDER)
    if f.endswith(".csv")
]

print("Existing CSV files:", len(existing_files))

for file in sorted(existing_files):
    print(file)
```

```python
existing_stock_names = {
    os.path.splitext(file)[0]
    for file in existing_files
}

stocks_df["File_Name"] = stocks_df["Ticker"].str.replace(
    ".NS", "", regex=False
)

stocks_to_download = stocks_df[
    ~stocks_df["File_Name"].isin(existing_stock_names)
].copy()

print("Total stocks:", len(stocks_df))
print("Already available:", len(existing_stock_names))
print("Still need to download:", len(stocks_to_download))
```

```python
import yfinance as yf
import pandas as pd
import os
import time

successful_downloads = []
failed_downloads = []

START_DATE = "2015-01-01"
END_DATE = None   # Downloads data up to the latest available date

for index, row in stocks_to_download.iterrows():

    ticker = row["Ticker"]
    company = row["Company"]
    file_name = row["File_Name"]

    print(f"\nDownloading {index + 1}/{len(stocks_df)}: {ticker} - {company}")

    try:
        # Download stock data
        df = yf.download(
            ticker,
            start=START_DATE,
            end=END_DATE,
            progress=False,
            auto_adjust=False
        )

        # Check if data was downloaded
        if df.empty:
            print(f"❌ No data found for {ticker}")
            failed_downloads.append(ticker)
            continue

        # Reset index so Date becomes a column
        df.reset_index(inplace=True)

        # Save CSV
        save_path = os.path.join(DATA_FOLDER, f"{file_name}.csv")
        df.to_csv(save_path, index=False)

        successful_downloads.append(ticker)

        print(f"✅ Saved: {file_name}.csv | Records: {len(df)}")

        # Small delay to avoid too many requests
        time.sleep(1)

    except Exception as e:
        print(f"❌ Failed: {ticker}")
        print("Error:", e)

        failed_downloads.append(ticker)

        time.sleep(2)


print("\n" + "=" * 50)
print("BATCH DOWNLOAD COMPLETED")
print("=" * 50)

print("Successful downloads:", len(successful_downloads))
print("Failed downloads:", len(failed_downloads))

if failed_downloads:
    print("\nFailed tickers:")
    print(failed_downloads)
```

```python
csv_files = [
    f for f in os.listdir(DATA_FOLDER)
    if f.endswith(".csv")
]

print("Total CSV files available:", len(csv_files))
print()

# Convert filenames into stock symbols
available_stocks = set(
    os.path.splitext(f)[0] for f in csv_files
)

# Expected stock symbols from stocks.csv
expected_stocks = set(
    stocks_df["Ticker"]
    .str.replace(".NS", "", regex=False)
)

missing_stocks = expected_stocks - available_stocks
extra_stocks = available_stocks - expected_stocks

print("Expected stocks:", len(expected_stocks))
print("Available CSV files:", len(available_stocks))
print("Missing stocks:", len(missing_stocks))
print("Extra stocks:", len(extra_stocks))

if len(missing_stocks) > 0:
    print("\nMissing:")
    print(sorted(missing_stocks))

if len(extra_stocks) > 0:
    print("\nExtra:")
    print(sorted(extra_stocks))
```

```python
sample_file = os.path.join(DATA_FOLDER, "ACC.csv")

df = pd.read_csv(sample_file)

print("Before cleaning:")
display(df.head())
```

```python
# Remove rows where Date is missing
df = df.dropna(subset=["Date"])

# Convert Date to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Convert price columns to numeric
price_columns = ["Adj Close", "Close", "High", "Low", "Open", "Volume"]

for col in price_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove any rows with invalid Close price
df = df.dropna(subset=["Close"])

# Sort by date
df = df.sort_values("Date").reset_index(drop=True)

print("After cleaning:")
print("Shape:", df.shape)

display(df.head())
```
