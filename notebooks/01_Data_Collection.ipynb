```python
RAW_DATA_PATH = r"D:\projects\AI-Stock-Analytics-System\data\raw"

successful_downloads = []
failed_downloads = []

for i, ticker in enumerate(stock_list, start=1):

    file_name = ticker.replace(".NS", "") + ".csv"
    file_path = os.path.join(RAW_DATA_PATH, file_name)

    # Skip if file already exists
    if os.path.exists(file_path):
        print(f"[{i}/{len(stock_list)}] {file_name} already exists. Skipping...")
        continue

    try:
        print(f"[{i}/{len(stock_list)}] Downloading {ticker}...")

        df = yf.download(
            ticker,
            start=START_DATE,
            end=END_DATE,
            progress=False
        )

        if df.empty:
            print(f"No data found for {ticker}")
            failed_downloads.append(ticker)
            continue

        df.to_csv(file_path)

        successful_downloads.append(ticker)

        print(f"Saved {file_name}")

    except Exception as e:
        print(f"Error downloading {ticker}: {e}")
        failed_downloads.append(ticker)
```

```python
print(f"Downloaded: {len(successful_downloads)}")
print(f"Failed: {len(failed_downloads)}")
```

```python
for ticker in failed_downloads:
    print(f"Retrying {ticker}")

    df = yf.download(
        ticker,
        start=START_DATE,
        end=END_DATE,
        progress=False
    )

    print(df.head())
```
